```python
 import os
 import glob
 import shutil
 from ultralytics import YOLO
 from roboflow import Roboflow
 print("Mengunduh dataset dari Roboflow")
 dataset_yaml_path=None
 dataset_location=None
 try:
  rf = Roboflow(api_key="epv0F946ZB0aFwaP7yL2")
  project = rf.workspace("afifulislam").project("revised-thesis-dataset")
  version = project.version(4)
  dataset = version.download("yolov12")
  dataset_location=dataset.location
  dataset_yaml_path=os.path.join(dataset_location,"data.yaml")
  print(f"Dataset berhasil diunduh ke:{dataset_location}")
  print(f"Pathfiledata.yaml:{dataset_yaml_path}")
except:
  print("Pelatihan tidak dapat dilanjutkan karena dataset.yaml tidak ditemukan atau gagal diunduh.")
```

```python
if dataset_yaml_path and os.path.exists(dataset_yaml_path):
  PRETRAINED_MODEL = "yolo12n.pt"
  try:
      model = YOLO(PRETRAINED_MODEL)
      print(f"Model {PRETRAINED_MODEL} berhasil dimuat.")

      EPOCHS = 200
      IMG_SIZE = 640
      BATCH_SIZE = 64
      LEARNING_RATE = 0.0013

      PROJECT_NAME = "Deteksi_Kendaraan_Bermotor"
      RUN_NAME = f"{PRETRAINED_MODEL.split('.')[0]}_DeteksiKendaraan_{version.version}_{EPOCHS}epochs"

      print("\nMemulai pelatihan model!")
      print(f"Dataset: {dataset_yaml_path}")
      print(f"Pretrained Model: {PRETRAINED_MODEL}")
      print(f"Epochs: {EPOCHS}")
      print(f"Image Size: {IMG_SIZE}")
      print(f"Batch Size: {BATCH_SIZE}")

      results = model.train(
          data=dataset_yaml_path,
          epochs=EPOCHS,
          imgsz=IMG_SIZE,
          batch=BATCH_SIZE,
          project=PROJECT_NAME,
          name=RUN_NAME,
          patience=0,
          optimizer="SGD",
          lr0=LEARNING_RATE,
          momentum=0.949,
          # ==================================
          exist_ok=True
      )

      print("Pelatihan selesai!")

      print(f"Hasil pelatihan disimpan di direktori: ./{PROJECT_NAME}/{RUN_NAME}")
      best_model_path = os.path.join(PROJECT_NAME,
      RUN_NAME, "weights", "best.pt")
      print(f"Model terbaik disimpan di:{best_model_path}")
  except Exception as e:
      print(f"Terjadi kesalahan saat melatih model:{e}")
      print("Penyebab umum: path data.yaml salah, format dataset tidak benar, VRAM GPU kurang.")
else:
  print("Pelatihan tidak dapat dilanjutkan karena dataset.yaml tidak ditemukan atau gagal diunduh.")
```
